# Clustering — PyCaret 4.0

Unsupervised: no `target`, no `compare_models` / `tune_model`. You pick an algorithm with `create_model`, then call `assign_model` to label every row.

```python
# PyCaret 4.0 — OOP-only engine
# https://github.com/pycaret/pycaret
import warnings
warnings.filterwarnings("ignore")
import pycaret
print("PyCaret", pycaret.__version__)
```

## 1. Load data

```python
from pycaret.datasets import get_data
data = get_data("jewellery")
data.head()
```

## 2. Fit the experiment

```python
from pycaret.tasks import ClusteringExperiment

exp = ClusteringExperiment(session_id=42, n_jobs=1).fit(data)
exp.X.shape
```

## 3. Create a clustering model

`create_model` returns a `CreateResult` with the fitted `.pipeline` and CV metrics.

```python
km = exp.create_model("kmeans", num_clusters=4)
km.metrics
```

## 4. Assign cluster labels to every row

```python
labelled = exp.assign_model(km.pipeline)
labelled.head()
```

## 5. Predict clusters on new data

```python
new_data = data.sample(5, random_state=42)
exp.predict_model(km.pipeline, data=new_data).predictions
```
