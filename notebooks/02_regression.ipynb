# Regression — PyCaret 4.0

Predict a continuous target. Same shape as the classification notebook, different `Experiment` subclass.

```python
# PyCaret 4.0 — OOP-only engine
# https://github.com/pycaret/pycaret
import warnings
warnings.filterwarnings("ignore")
import pycaret
print("PyCaret", pycaret.__version__)
```

## 1. Load data

```python
from pycaret.datasets import get_data
data = get_data("boston")
data.head()
```

## 2. Fit an experiment

```python
from pycaret.tasks import RegressionExperiment

exp = RegressionExperiment(target="medv", session_id=42, n_jobs=1).fit(data)
exp.X_train.shape, exp.X_test.shape
```

## 3. Compare models

```python
result = exp.compare_models(include=["lr", "ridge", "rf"])
result.leaderboard
```

## 4. Predict on the test set

```python
preds = exp.predict_model(result.best)
preds.predictions.head()
```

## 5. Persist

```python
from pycaret import save_model
save_model(result.best, "artifacts/boston_regressor")
```
