# Time-series forecasting — PyCaret 4.0

Input is a univariate `pandas.Series` (or a DataFrame with a named target column). `fh` is the forecast horizon.

```python
# PyCaret 4.0 — OOP-only engine
# https://github.com/pycaret/pycaret
import warnings
warnings.filterwarnings("ignore")
import pycaret
print("PyCaret", pycaret.__version__)
```

## 1. Load data

```python
from pycaret.datasets import get_data
y = get_data("airline")
y.head()
```

## 2. Fit the experiment

`fh=12` means forecast 12 steps ahead.

```python
from pycaret.tasks import TimeSeriesExperiment

exp = TimeSeriesExperiment(fh=12, session_id=42, n_jobs=1).fit(y)
```

## 3. Compare forecasters

```python
result = exp.compare_models(include=["naive", "arima", "exp_smooth"])
result.leaderboard
```

## 4. Forecast

```python
best = result.best
forecast = exp.predict_model(best)
forecast.predictions
```
