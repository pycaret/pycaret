# Anomaly detection — PyCaret 4.0

Unsupervised: `create_model` picks the algorithm, `assign_model` labels every row with an anomaly score.

```python
# PyCaret 4.0 — OOP-only engine
# https://github.com/pycaret/pycaret
import warnings
warnings.filterwarnings("ignore")
import pycaret
print("PyCaret", pycaret.__version__)
```

## 1. Load data

```python
from pycaret.datasets import get_data
data = get_data("anomaly")
data.head()
```

## 2. Fit the experiment

```python
from pycaret.tasks import AnomalyExperiment

exp = AnomalyExperiment(session_id=42, n_jobs=1).fit(data)
exp.X.shape
```

## 3. Create an anomaly-detection model

```python
iforest = exp.create_model("iforest")
iforest.params
```

## 4. Label every row

```python
labelled = exp.assign_model(iforest.pipeline)
labelled.head()
```

Rows with `Anomaly == 1` are flagged; `Anomaly_Score` is the raw score.
